```python
def traductor(z):
    return 1 / (1 + np.exp(-z))

def neurona(lecturas, pesos, sesgo):
    suma = 0
    for lectura, peso in zip(lecturas, pesos):
        suma = suma + lectura * peso
    suma = suma + sesgo
    return traductor(suma)
```

```python
import math

def neurona(entradas, pesos, sesgo):
    suma = sum(x * w for x, w in zip(entradas, pesos)) + sesgo
    return 1 / (1 + math.exp(-suma))
```

```python
probabilidad = neurona([3, 2, 1], [0.8, 0.5, 0.1], -2)

print("Probabilidad de falla:", round(probabilidad * 100), "%")
```
